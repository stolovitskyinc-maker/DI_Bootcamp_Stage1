```python
# Exam 1: Number Guessing Game 🎲 (Mandatory)
# Build a fun Number Guessing Game in Python! 🐍 The program picks a random number between 1-100, and you have 7 attempts to guess it. Get hints if you’re too high 📈 or too low 📉! Perfect for practicing loops 🔄, conditionals ❓, and user input ⌨️.

import random

def number_guessing_game():
    # Pick a random number between 1 and 100 inclusive
    secret_number = random.randint(1, 100)
    max_attempts = 7

    print("Welcome to the Number Guessing Game! 🎲")
    print("I am thinking of a number between 1 and 100.")
    print(f"You have {max_attempts} attempts to guess it correctly. Good luck! 🐍\n")

    # Loop through the allowable attempts
    for attempt in range(1, max_attempts + 1):
        user_input = input(f"Attempt {attempt}/{max_attempts} - Enter your guess: ")

        # Validate that the user input is a positive integer
        if not user_input.isdigit():
            print("❌ Invalid input! Please enter a whole number between 1 and 100.\n")
            continue

        guess = int(user_input)

        # Check if the guess matches, or guide the player with hints
        if guess == secret_number:
            print(f"🎉 Congratulations! You guessed the correct number {secret_number} in {attempt} attempts!")
            break
        elif guess < secret_number:
            print("📉 Too low! Try guessing a higher number. 📈\n")
        else:
            print("📈 Too high! Try guessing a lower number. 📉\n")

    else:
        # Executed if the loop completes fully without breaking (running out of attempts)
        print(f"💥 Game Over! You've used all 7 attempts. The secret number was {secret_number}.")

if __name__ == "__main__":
    number_guessing_game()
```
